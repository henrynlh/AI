## PEAS: Máy hút bụi
### P: Là khả năng tối ưu nhất để sàn nhà sạch và tiết kiệm năng lượng nhất
### E: Là ma trận mxn biểu thị các số 0 và 1; 0 là sàn sạch và 1 là sàn bẩn tại vị trí của máy hút bụi
### A: Các hàm để agent hiểu và hành động (tập luật)
### S: Cơ chế, hàm để agent đọc dữ liệu (vị trí máy hút bụi, trạng thái,...)

```python
import random
```

```python
def input_floor():
    m = int(input("Nhập số dòng (m): "))
    n = int(input("Nhập số cột (n): "))
    print(f"Nhập ma trận {m}x{n}, 0 = sạch, 1 = bẩn:")

    floor = []
    for i in range(m):
        while True:
            row_input = input(f"Dòng {i + 1}: ")
            row = row_input.split()
            if len(row) != n:
                print(f"Phải nhập đúng {n} số. Nhập lại.")
                continue
            try:
                row = [int(x) for x in row]
            except ValueError:
                print("Chỉ được nhập số 0 hoặc 1. Nhập lại.")
                continue
            if any(x not in (0, 1) for x in row):
                print("Chỉ được nhập 0 hoặc 1. Nhập lại.")
                continue
            floor.append(row)
            break

    return floor, m, n
```

```python
def get_possible_moves(pos, m, n):
    x, y = pos
    moves = []
    if x > 0: moves.append("up")
    if x < m-1: moves.append("down")
    if y > 0: moves.append("left")
    if y < n-1: moves.append("right")
    return moves
```

```python
def apply_move(pos, move):
    x, y = pos
    if move == "up":
        x -= 1
    elif move == "down":
        x += 1
    elif move == "left":
        y -= 1
    elif move == "right":
        y += 1
    return (x, y)
```

```python
def print_floor(floor):
    for row in floor:
        print(row)
```

```python
def random_clean_prioritize_dirty(floor, m, n):
    pos = (random.randint(0, m-1), random.randint(0, n-1))
    step = 0

    print("\nTrạng thái ban đầu:")
    print_floor(floor)
    print("Vị trí ban đầu của máy hút bụi:", pos)

    x0, y0 = pos
    if floor[x0][y0] == 1:
        floor[x0][y0] = 0
        print(f"\nVị trí ban đầu bẩn ({x0},{y0}) → Đã hút bụi")
        print("Trạng thái hiện tại sau khi hút vị trí ban đầu:")
        print_floor(floor)

    while True:
        step += 1
        print(f"\n========== LẦN ROLL {step} ==========")

        possible_moves = get_possible_moves(pos, m, n)

        dirty_moves = []
        for move in possible_moves:
            new_pos = apply_move(pos, move)
            if floor[new_pos[0]][new_pos[1]] == 1:
                dirty_moves.append(move)

        if dirty_moves:
            selected_move = random.choice(dirty_moves)
            print("Ưu tiên đi hướng có ô bẩn:", selected_move)
        else:
            selected_move = random.choice(possible_moves)
            print("Không có ô bẩn xung quanh, đi random:", selected_move)

        pos = apply_move(pos, selected_move)
        x, y = pos

        if floor[x][y] == 1:
            floor[x][y] = 0
            print(f"Đã hút bụi tại ô ({x},{y})")
        else:
            print(f"Ô ({x},{y}) đã sạch")

        print("Trạng thái hiện tại:")
        print_floor(floor)

        if all(num == 0 for row in floor for num in row):
            print("\nTẤT CẢ CÁC Ô ĐÃ SẠCH! Hoàn thành.")
            break

    print("\nTổng số lần roll:", step)
```

```python
floor, m, n = input_floor()
random_clean_prioritize_dirty(floor, m, n)
```
